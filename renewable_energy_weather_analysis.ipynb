<div style="font-size: 18px;font-family:Times New Roman">Dans un monde où la transition vers une énergie plus propre et renouvelable est devenue une priorité, la compréhension de la relation entre les conditions météorologiques et la production d'énergie renouvelable revêt d'une importance cruciale. Les fluctuations du rayonnement solaire, de la température, de l'humidité et des précipitations peuvent avoir un impact significatif sur la disponibilité et l'efficacité des sources d'énergie renouvelable telles que l'énergie solaire, éolienne et hydraulique.</div>

```python
from sklearn.feature_selection import mutual_info_regression
from sklearn.preprocessing import LabelEncoder

# Variables qualitative
varQualitatif2 = ['isSun', 'weather_type']

# Encodage des variables qualitatives
encoder = LabelEncoder()
df_encoded = df.copy()
for var in varQualitatif2:
    df_encoded[var] = encoder.fit_transform(df[var])

# Calculer l'information mutuelle entre chaque variable qualitative et la variable cible "Energy delta[Wh]"
mi_energy_delta = []
for var in varQualitatif2:
    X = df_encoded[[var]]
    y = df['Energy delta[Wh]']
    mi = mutual_info_regression(X, y)
    mi_energy_delta.append(mi[0])  # Nous prenons la valeur de l'information mutuelle pour chaque variable
```

```python
plt.figure(figsize=(10, 5))
# Tracer les informations mutuelles
plt.bar(varQualitatif2, mi_energy_delta, color='blue')
plt.title('Information Mutuelle entre les variables qualitatives et Energy delta[Wh]')
plt.xlabel('Variables Qualitatives')
plt.ylabel('Information Mutuelle')
plt.grid(True)
# Afficher le graphique
plt.xticks(rotation=45)
plt.show()
```

```python
def createStatTable1(data, colName, dropna = False, sort = False):
    c1 = data[colName].value_counts(sort=sort, dropna= dropna)
    p1 = data[colName].value_counts(sort=sort, normalize=True, dropna= dropna)
    df1 = pd.concat((c1,p1), axis=1)
    df1.columns = ['Count', 'Frequency']
    df1['Cumulative count'] = df1['Count'].cumsum()
    df1['Cumulative frequency'] = df1['Frequency'].cumsum()
    return df1
```

```python
def createStatTable2(df, variable, bins):
    # Définir les intervalles et les étiquettes
    intervalles = pd.cut(df[variable], bins=bins)
    
    # Calculer l'effectif pour chaque intervalle
    effectifs = intervalles.value_counts().reset_index()
    effectifs.columns = ['Modalité', 'Effectif']

    # Calculer l'effectif cumulé
    effectifs['Effectif cumulé'] = effectifs['Effectif'].cumsum()

    # Calculer la fréquence pour chaque intervalle
    effectifs['Fréquence'] = effectifs['Effectif'] / len(df)

    # Calculer la fréquence cumulée croissante
    effectifs['Fréquence cumulée'] = effectifs['Fréquence'].cumsum()

    return effectifs
```

```python
print('Table pour le type de météo ')
createStatTable1(df, 'weather_type')
```

```python
print("Table de la quantité de rayonnement solaire reçue par une surface horizontale")
statistiques = createStatTable2(df, 'GHI', bins=[-1, 5, 10, 100, 200, 3000])
print(statistiques)
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Sélectionner les colonnes numériques et normaliser les données
X_scaled = StandardScaler().fit_transform(df_quantitatif)

# Effectuer l'ACP
pca = PCA()
pca.fit(X_scaled)

# Composantes principales
principal_components = pca.components_

# Variance expliquée par chaque composante principale
explained_variance = pca.explained_variance_ratio_
```

```python
plt.figure(figsize=(10, 4))
ratiosCR = pca.explained_variance_ratio_
print(ratiosCR)
plt.bar(range(len(ratiosCR)), ratiosCR)
plt.xticks(range(len(ratiosCR)))
plt.xlabel("Composante principale")
plt.ylabel("% de variance expliquée")
plt.show()
```
